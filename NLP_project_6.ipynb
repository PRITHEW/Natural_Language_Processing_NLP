```python
import os
import pandas as pd

os.environ["HF_HUB_ENABLE_HF_INFO"] = "False"

# Create dummy dataframes instead of loading from datasets
# Replace this with your actual data
train_data = {
    'text': [
        "India wins cricket world cup!",
        "Tesla announces new electric car model.",
        "New medical breakthrough in cancer research.",
        "Stock market surges on positive economic news.",
        "Scientists discover new exoplanet."
    ],
    'label': [1, 2, 3, 0, 3] # Example labels: 0: World, 1: Sports, 2: Business, 3: Sci/Tech
}
test_data = {
    'text': [
        "Local team loses close match.",
        "Company reports record profits.",
        "New findings in quantum physics."
    ],
    'label': [1, 2, 3]
}

train_df = pd.DataFrame(train_data)
test_df = pd.DataFrame(test_data)

print(train_df.head())
```

```python
print(train_df.shape)
print(train_df['label'].value_counts())
```

```python
from nltk.corpus import stopwords
import string
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))

def clean_text(text):

    text = text.lower()

    text = text.translate(str.maketrans('', '', string.punctuation))

    words = word_tokenize(text)

    words = [word for word in words if word not in stop_words]

    return " ".join(words)

train_df["clean_text"] = train_df["text"].apply(clean_text)
test_df["clean_text"] = test_df["text"].apply(clean_text)

train_df.head()
```

```python
from datasets import load_dataset
import pandas as pd
import nltk
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000)

X_train = tfidf.fit_transform(train_df["clean_text"])
X_test = tfidf.transform(test_df["clean_text"])

y_train = train_df["label"]
y_test = test_df["label"]
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))

print(classification_report(y_test, y_pred))
```

```python
news = [
    "India defeated Australia by six wickets in the cricket match."
]

news_clean = [clean_text(i) for i in news]

news_vector = tfidf.transform(news_clean)

prediction = model.predict(news_vector)

labels = {
    0:"World",
    1:"Sports",
    2:"Business",
    3:"Science/Technology"
}

print("Predicted Category:", labels[prediction[0]])
```
